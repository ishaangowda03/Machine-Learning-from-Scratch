```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import make_blobs
```

```python
X , y = make_blobs(n_samples=2000 , n_features=2, centers=3, cluster_std=3, random_state=43)
n_features = X.shape[1]
m = X.shape[0]
print(X.shape,y.shape)
```

```python
#Visualizing the Data
xt = np.array([2,2])
for i in range(m):
    if y[i]==0:
        plt.scatter(X[i,0],X[i,1],c='r',label="red")
    elif y[i]==1:
        plt.scatter(X[i,0],X[i,1],c='g',label="green")
    else:
        plt.scatter(X[i,0],X[i,1],c='b',label="blue")
plt.show()
```

```python
#Normalise the Data
def Normalise(X):
    mu = X.mean(axis=0)
    std = X.std(axis=0)
    return (X-mu)/std
X = Normalise(X)
```

```python
def Distance(p,q):
    return np.sqrt(np.sum((p-q)**2))
```

```python
def KNN(X,y,xt,k=5):
    dlist = []
    m = X.shape[0]
    for i in range(m):
        dist = Distance(xt,X[i])
        dlist.append((dist,y[i]))
    dlist = sorted(dlist)
    dlist = np.array(dlist[:k])
    labels = dlist[:,1]

    labels,count = np.unique(labels,return_counts=True)
    idx = count.argmax()
    prediction = labels[idx]
    print(int(prediction))
    return int(prediction)
```

```python
KNN(X, y, xt, k=200)
```
