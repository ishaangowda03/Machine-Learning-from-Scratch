```python
import pandas as pd
import numpy as np
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
```

```python
data = pd.read_csv('train.csv')
X = data.iloc[ : , : -1]
y = data.iloc[ : , -1]
```

```python
#Normalise the Data
mu = X.mean(axis=0)
std = X.std(axis = 0)
X = (X-mu)/std
```

```python
best_features = SelectKBest(score_func=f_classif)
fit = best_features.fit(X,y)
print(fit.scores_)
```

```python
fitScores = pd.DataFrame(fit.scores_)
fitColumns = pd.DataFrame(X.columns)
featuresScores = pd.concat([fitColumns, fitScores], axis = 1)
featuresScores.columns = ['Features','Scores']
featuresScores = featuresScores.sort_values(by = 'Scores', ascending = False)
top_10 = list(featuresScores[:10]['Features'].values)
```

```python
model = LogisticRegression()
print(model.fit(X[top_10],y))
scores = cross_val_score(model, X[top_10], y, cv = 10)
print(scores.mean())
```
